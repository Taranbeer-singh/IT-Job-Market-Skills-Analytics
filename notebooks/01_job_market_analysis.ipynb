```python
top_jobs = df["Job_Title"].value_counts().head(10)

top_jobs
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

top_jobs.plot(kind="bar")

plt.title("Top 10 Most Common IT Job Roles")
plt.xlabel("Job Role")
plt.ylabel("Number of Job Listings")

plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

ax = top_jobs.plot(kind="bar")

plt.title("Top 10 Most Common IT Job Roles")
plt.xlabel("Job Role")
plt.ylabel("Number of Job Listings")

plt.xticks(rotation=45, ha="right")

for i, value in enumerate(top_jobs):
    ax.text(i, value + 5, str(value), ha="center")

plt.tight_layout()
plt.show()
```

```python
salary_by_experience = df.groupby("Experience_Level")["Salary_LPA"].mean().sort_values()

salary_by_experience
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 5))

ax = salary_by_experience.plot(kind="bar")

plt.title("Average Salary by Experience Level")
plt.xlabel("Experience Level")
plt.ylabel("Average Salary (LPA)")

for i, value in enumerate(salary_by_experience):
    ax.text(i, value + 1, f"{value:.1f}", ha="center")

plt.tight_layout()
plt.show()
```

```python
top_skills = all_skills.value_counts().head(10)

top_skills
```

```python
plt.figure(figsize=(10, 6))

ax = top_skills.sort_values().plot(kind="barh")

plt.title("Top 10 Most Frequently Required Technical Skills")
plt.xlabel("Number of Job Listings")
plt.ylabel("Technical Skill")

for i, value in enumerate(top_skills.sort_values()):
    ax.text(value + 20, i, str(value), va="center")

plt.tight_layout()
plt.show()
```

```python
salary_by_role = df.groupby("Job_Title")["Salary_LPA"].mean()
salary_by_role = salary_by_role.sort_values(ascending=False).head(10)
salary_by_role
```

```python
plt.figure(figsize=(10, 6))

ax = salary_by_role.sort_values().plot(kind="barh")

plt.title("Top 10 Highest Paying IT Job Roles")
plt.xlabel("Average Salary (LPA)")
plt.ylabel("Job Role")

for i, value in enumerate(salary_by_role.sort_values()):
    ax.text(value + 0.3, i, f"{value:.1f}", va="center")

plt.tight_layout()
plt.show()
```

```python
salary_by_work_mode = (
    df.groupby("Work_Mode")["Salary_LPA"]
    .mean()
    .sort_values()
)

salary_by_work_mode
```

```python
plt.figure(figsize=(8, 5))

ax = salary_by_work_mode.plot(kind="bar")

plt.title("Average Salary by Work Mode")
plt.xlabel("Work Mode")
plt.ylabel("Average Salary (LPA)")

for i, value in enumerate(salary_by_work_mode):
    ax.text(i, value + 0.2, f"{value:.1f}", ha="center")

plt.tight_layout()
plt.show()
```
