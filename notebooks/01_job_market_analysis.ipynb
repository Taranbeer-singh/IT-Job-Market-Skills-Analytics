```python
import pandas as pd

df = pd.read_csv("../data/india_job_market_2024_2026.csv")

df.head()
```

```python
print("Rows and Columns:", df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)
```

```python
missing_values = df.isnull().sum()

print(missing_values)
```

```python
print("Total duplicate rows:", df.duplicated().sum())
```

```python
print("Salary Range:", df["Salary_LPA"].min(), "to", df["Salary_LPA"].max())

print("Openings Range:", df["Openings"].min(), "to", df["Openings"].max())

print("Applicants Range:", df["Applicants"].min(), "to", df["Applicants"].max())

print("Company Rating Range:", df["Company_Rating"].min(), "to", df["Company_Rating"].max())
```

```python
categorical_columns = [
    "Job_Title",
    "Company_Type",
    "Industry",
    "Location_Tier",
    "Experience_Level",
    "Job_Type",
    "Work_Mode",
    "Education_Required"
]

for column in categorical_columns:
    print(f"\n--- {column} ---")
    print(df[column].value_counts())
```

```python
for column in [
    "Company_Type",
    "Industry",
    "Location_Tier",
    "Experience_Level",
    "Job_Type",
    "Work_Mode",
    "Education_Required"
]:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
print("Job Type:")
print(df["Job_Type"].value_counts())

print("\nWork Mode:")
print(df["Work_Mode"].value_counts())
```

```python
df["Date_Posted"] = pd.to_datetime(df["Date_Posted"])

print(df["Date_Posted"].dtype)
print(df["Date_Posted"].min())
print(df["Date_Posted"].max())
```
